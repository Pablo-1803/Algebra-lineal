```python
import random
import numpy as np
import ipywidgets as widgets
from IPython.display import display, clear_output

class TicTacToeMatrix:
    """
    Clase que representa el motor matemático del juego utilizando conceptos de Álgebra Lineal.

    Representación del estado en el espacio vectorial R^(3x3):
      0  -> Casilla vacía
      1  -> Jugador Humano (X)
     -1  -> CPU (O)
    """
    def __init__(self):
        # Inicializamos el tablero como una matriz nula de 3x3 (elemento neutro de la adición de matrices)
        self.board = np.zeros((3, 3), dtype=int)

    def restart(self):
        """Reinicia la matriz a su estado inicial nulo."""
        self.board.fill(0)

    def get_empty_cells(self):
        """
        Retorna las posiciones (coordenadas) vacías en la matriz.
        Utiliza operaciones de indexación booleana sobre la matriz.
        """
        rows, cols = np.where(self.board == 0)
        return list(zip(rows, cols))

    def make_move(self, row, col, player):
        """Asigna el escalar del jugador (1 o -1) en la coordenada de la matriz."""
        if self.board[row, col] == 0:
            self.board[row, col] = player
            return True
        return False

    def check_winner(self):
        """
        Determina si existe un ganador utilizando operaciones de Álgebra Lineal.

        Lógica matemática:
        - Sean v_i las filas de la matriz y u_j sus columnas.
        - La suma de los componentes de un vector fila i es np.sum(A[i, :]). Si es igual a 3 (3 * 1) gana Humano; si es -3 (3 * -1) gana CPU.
        - La traza de la matriz (np.trace(A)) representa la suma de la diagonal principal.
        - La traza de la matriz reflejada horizontalmente (np.trace(np.fliplr(A))) representa la diagonal secundaria.
        """
        # 1. Suma de vectores fila (dimensión de reducción 1)
        row_sums = np.sum(self.board, axis=1)
        # 2. Suma de vectores columna (dimensión de reducción 0)
        col_sums = np.sum(self.board, axis=0)
        # 3. Traza de la diagonal principal
        diag_sum = np.trace(self.board)
        # 4. Traza de la diagonal secundaria (reflejando la matriz de izquierda a derecha)
        antidiag_sum = np.trace(np.fliplr(self.board))

        # Agrupamos todas las combinaciones lineales de interés (vías de victoria)
        all_paths = np.concatenate([row_sums, col_sums, [diag_sum, antidiag_sum]])

        if 3 in all_paths:
            return 1  # Humano gana
        elif -3 in all_paths:
            return -1 # CPU gana
        elif len(self.get_empty_cells()) == 0:
            return 0  # Empate (Espacio de juego lleno sin ganadores)

        return None # El juego continúa


class DecisionAgent:
    """
    Agente de toma de decisiones (CPU) con lógica heurística y matricial.
    """
    def __init__(self, difficulty="Difícil"):
        self.difficulty = difficulty

    def get_move(self, game_matrix):
        empty_cells = game_matrix.get_empty_cells()
        if not empty_cells:
            return None

        if self.difficulty == "Fácil":
            return self._random_move(empty_cells)

        elif self.difficulty == "Medio":
            # 50% probabilidad aleatorio, 50% inteligente
            if random.random() < 0.5:
                return self._random_move(empty_cells)
            else:
                return self._heuristic_move(game_matrix, empty_cells)

        else: # Difícil
            return self._heuristic_move(game_matrix, empty_cells)

    def _random_move(self, empty_cells):
        """Selección aleatoria simple sobre el conjunto de coordenadas libres."""
        return random.choice(empty_cells)

    def _heuristic_move(self, game_matrix, empty_cells):
        """
        Aplica una heurística basada en la simulación de estados futuros en la matriz.
        """
        # a) Simular movimiento de la CPU (-1) para buscar la victoria inmediata
        for r, c in empty_cells:
            # Clonamos el estado de la matriz temporalmente
            temp_board = np.copy(game_matrix.board)
            temp_board[r, c] = -1
            # Evaluamos de forma local la condición de victoria para este estado simulado
            if self._eval_simulated_win(temp_board) == -1:
                return (r, c)

        # b) Simular movimiento del Humano (1) para bloquear su posible victoria inmediata
        for r, c in empty_cells:
            temp_board = np.copy(game_matrix.board)
            temp_board[r, c] = 1
            if self._eval_simulated_win(temp_board) == 1:
                return (r, c)

        # c) Ocupar la posición central (1, 1) que algebraicamente domina las diagonales y centro
        if (1, 1) in empty_cells:
            return (1, 1)

        # d) Ocupar una esquina al azar si está libre
        corners = [(0, 0), (0, 2), (2, 0), (2, 2)]
        available_corners = [c for c in corners if c in empty_cells]
        if available_corners:
            return random.choice(available_corners)

        # e) Ocupar cualquier otra casilla restante
        return random.choice(empty_cells)

    def _eval_simulated_win(self, board_matrix):
        """Evalúa un estado matricial simulado sin alterar el juego real."""
        row_sums = np.sum(board_matrix, axis=1)
        col_sums = np.sum(board_matrix, axis=0)
        diag_sum = np.trace(board_matrix)
        antidiag_sum = np.trace(np.fliplr(board_matrix))
        all_paths = np.concatenate([row_sums, col_sums, [diag_sum, antidiag_sum]])

        if 3 in all_paths:
            return 1
        elif -3 in all_paths:
            return -1
        return None


class TicTacToeColabGUI:
    """
    Clase encargada de renderizar la interfaz gráfica interactiva en Google Colab
    usando widgets de IPython.
    """
    def __init__(self):
        self.game = TicTacToeMatrix()
        self.agent = DecisionAgent(difficulty="Difícil")
        self.buttons = {}
        self.output_area = widgets.Output()
        self.status_label = widgets.Label(value="¡Tu turno! Elige una casilla (X).")
        self.setup_ui()

    def setup_ui(self):
        # Selector de dificultad
        self.diff_dropdown = widgets.Dropdown(
            options=['Fácil', 'Medio', 'Difícil'],
            value='Difícil',
            description='Dificultad:',
            disabled=False
        )
        self.diff_dropdown.observe(self.on_difficulty_change, names='value')

        # Botón para reiniciar el juego
        self.reset_button = widgets.Button(
            description='Reiniciar Juego',
            button_style='warning',
            icon='refresh'
        )
        self.reset_button.on_click(self.on_reset_clicked)

        # Grid de 3x3 para el tablero
        grid_layout = widgets.GridspecLayout(3, 3, width='240px', height='240px')
        for r in range(3):
            for c in range(3):
                btn = widgets.Button(
                    description='',
                    layout=widgets.Layout(width='70px', height='70px'),
                    button_style='primary'
                )
                # Almacenamos las coordenadas del botón usando atributos dinámicos de Python
                btn.row = r
                btn.col = c
                btn.on_click(self.on_cell_clicked)
                grid_layout[r, c] = btn
                self.buttons[(r, c)] = btn

        # Contenedores
        control_box = widgets.HBox([self.diff_dropdown, self.reset_button])
        main_layout = widgets.VBox([
            widgets.HTML("<h2>Tres en Raya (Álgebra Lineal & POO)</h2>"),
            control_box,
            self.status_label,
            grid_layout,
            self.output_area
        ])

        clear_output()
        display(main_layout)

    def on_difficulty_change(self, change):
        self.agent.difficulty = change['new']

    def update_board_display(self):
        """Sincroniza el estado de los botones visuales con la matriz matemática de NumPy."""
        for (r, c), btn in self.buttons.items():
            val = self.game.board[r, c]
            if val == 1:
                btn.description = 'X'
                btn.button_style = 'success' # Verde para el Humano
                btn.disabled = True
            elif val == -1:
                btn.description = 'O'
                btn.button_style = 'danger' # Rojo para la CPU
                btn.disabled = True
            else:
                btn.description = ''
                btn.button_style = 'primary' # Azul
                btn.disabled = False

    def disable_all_cells(self):
        for btn in self.buttons.values():
            btn.disabled = True

    def on_reset_clicked(self, b):
        self.game.restart()
        # Limpiamos explícitamente el texto de todos los botones para evitar caracteres residuales
        for btn in self.buttons.values():
            btn.description = ''
            btn.button_style = 'primary'
            btn.disabled = False
        self.update_board_display()
        self.status_label.value = "¡Juego reiniciado! Tu turno (X)."
        with self.output_area:
            clear_output()

    def on_cell_clicked(self, btn):
        # Movimiento del Humano
        row, col = btn.row, btn.col
        if self.game.make_move(row, col, 1):
            self.update_board_display()

            # Verificar si el humano ganó con esta jugada
            state = self.game.check_winner()
            if self._handle_game_over(state):
                return

            # Turno de la CPU
            self.status_label.value = "La CPU está pensando..."
            cpu_move = self.agent.get_move(self.game)
            if cpu_move:
                self.game.make_move(cpu_move[0], cpu_move[1], -1)
                self.update_board_display()

                # Verificar si la CPU ganó
                state = self.game.check_winner()
                if self._handle_game_over(state):
                    return

            self.status_label.value = "Tu turno (X)."

    def _handle_game_over(self, state):
        if state is not None:
            self.disable_all_cells()
            if state == 1:
                self.status_label.value = "🎉 ¡Felicidades! Has vencido a la CPU."
            elif state == -1:
                self.status_label.value = "🤖 La CPU ha ganado. ¡Sigue intentándolo!"
            elif state == 0:
                self.status_label.value = "🤝 ¡Es un empate! Excelente partida."
            return True
        return False

# Ejecución e inicialización de la interfaz en Colab
tictactoe_app = TicTacToeColabGUI()
```
